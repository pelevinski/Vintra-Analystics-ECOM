```python
import pandas as pd
arquivo = 'vintra_base_dados.xlsx'
dados = pd.read_excel(arquivo, sheet_name=None)
df = dados['Vendas']
print(df.isnull().sum())
df['desconto_pct'] = df['desconto_pct'].fillna(0)
df = df.dropna(subset=['quantidade'])
df = df.drop_duplicates('id_pedido')
print(df.duplicated().sum())
dados['Vendas'] = df
df = dados['Clientes']
df['estado'] = df['estado'].str.upper()
print((df['estado'] != df['estado'].str.upper()).sum())
dados['Clientes'] = df
with pd.ExcelWriter('vintra_dados_tratado.xlsx') as writer:
    for nome_aba, df in dados.items():
        df.to_excel(writer, sheet_name=nome_aba, index=False)

from google.colab import files

files.download('vintra_dados_tratado.xlsx')
```
